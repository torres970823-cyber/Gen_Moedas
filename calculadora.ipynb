```python
# para criar uma funçao em python
# def nome_da_função (parametros) --- apos cria o def na outra linha dar tab
  #tab sua função
  #tab return resultado

# queremo criar uma função que é uma calculadora

# def nome_da_função (parametros)
def calculadora (num1, num2, operacao):

  if operacao == '+':
    return num1 + num2
  elif operacao == '-':
    return num1 - num2
  elif operacao == '*':
    return num1 * num2
  elif operacao == '/':
    if num2 == 0:
      return 'erro, nao pdoe dividir por zero (0)'
    else:
      return num1 / num2
  else:
    return 'escolha um operador valido (+, -, *ou /)'
```

```python
# vamos simular nossa calculadora

num1 = float(input('informe o primeiro numero\n'))
num2 = float(input('informe o segundo numero\n'))
operacao =input('informe um operador\n')

calculadora(num1, num2, operacao)
```
